```python
urls = {
    "2021-01": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2021-01.parquet",
    "2021-02": "https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2021-02.parquet",
}

for month, url in urls.items():
    destination = data_dir / f"yellow_tripdata_{month}.parquet"
    if not destination.exists():
        urlretrieve(url, destination)
```

```python
import pandas as pd

# 1. Rechercher tous les fichiers yellow_tripdata_*.parquet dans data
files = sorted(data_dir.glob("yellow_tripdata_*.parquet"))
frames = []

for parquet_file in files:
  # 2. Lire chaque fichier avec pandas.read_parquet
  df = pd.read_parquet(parquet_file)

  # 3. Extraire le mois depuis parquet_file.stem et ajouter la colonne source_month
  month = parquet_file.stem.removeprefix("yellow_tripdata_")
  df["source_month"] = month

  # 4. Ajouter le DataFrame à la liste
  frames.append(df)

# Rassembler tous les DataFrames avec pandas.concat
trips = pd.concat(frames, ignore_index=True)

# 5. Afficher le nombre de lignes pour chaque mois
print(trips["source_month"].value_counts().sort_index())
```
